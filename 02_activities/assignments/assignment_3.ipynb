#### **Question 2:** 
#### Data-visualization

Let's create plots to visualize the relationships between the features (e.g., alcohol, malic acid, ash...).


```python
def plot_feature_pairs(data, feature_names, color_labels=None, title_prefix=''):
    """
    Helper function to create scatter plots for all possible pairs of features.
    
    Parameters:
    - data: DataFrame containing the features to be plotted.
    - feature_names: List of feature names to be used in plotting.
    - color_labels: Optional. Cluster or class labels to color the scatter plots.
    - title_prefix: Optional. Prefix for plot titles to distinguish between different sets of plots.
    """
    # Create a figure for the scatter plots
    plt.figure(figsize=(60, 60))
    
    # Counter for subplot index
    plot_number = 1
    
    # Loop through each pair of features
    for i in range(len(feature_names)):
        for j in range(i + 1, len(feature_names)):
            plt.subplot(len(feature_names)-1, len(feature_names)-1, plot_number)
            
            # Scatter plot colored by labels if provided
            if color_labels is not None:
                plt.scatter(data[feature_names[i]], data[feature_names[j]], 
                            c=color_labels, cmap='viridis', alpha=0.7)
            else:
                plt.scatter(data[feature_names[i]], data[feature_names[j]], alpha=0.7)
            
            plt.xlabel(feature_names[i])
            plt.ylabel(feature_names[j])
            plt.title(f'{title_prefix}{feature_names[i]} vs {feature_names[j]}')
            
            # Increment the plot number
            plot_number += 1

    # Adjust layout to prevent overlap
    plt.tight_layout()

    # Show the plot
    plt.show()

# Get feature names
feature_names = wine_df.columns

# Use the helper function to plot scatter plots without coloring by cluster labels
plot_feature_pairs(wine_df, feature_names, title_prefix='Original Data: ')
```

#### **Question 3:** 
#### Data cleaning

```python
# Initialize the StandardScaler
scaler = StandardScaler()

# Scale all the features in the dataset
scaled_features = scaler.fit_transform(wine_df)

# Create a new DataFrame with scaled features
scaled_wine_df = pd.DataFrame(scaled_features, columns=wine_data.feature_names)

# Display the first few rows of the scaled DataFrame
print(scaled_wine_df.head())
```

Why is it important to standardize the features of a dataset before applying clustering algorithms like K-Means? Discuss the implications of using unstandardized data in your analysis.

My Answer: 

I standardized the data because the different features are not all measured on the same scale. Since K-Means uses the distance between data points to create clusters, a feature with much larger numerical values could have more influence on the clustering than the other features. Standardizing makes the scales more comparable so that the clustering is based more on the overall differences between the wine samples rather than mainly on whichever variables have the largest values.

#### **Question 4:** 
#### K-means clustering 

Apply the K-Means clustering algorithm to the Wine dataset. Choose the value 3 for the number of clusters (`k=3`) and fit the model. Assign cluster labels to the original data and add them as a new column in the DataFrame.

```python
# Your answer...
# Create a K-Means model with 3 clusters
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)

# Fit the model to the standardized data and get the cluster labels
cluster_labels = kmeans.fit_predict(scaled_wine_df)

# Copy the original data and add the cluster labels as a new column
clustered_wine_data = wine_df.copy()
clustered_wine_data["Cluster"] = cluster_labels


# Use the helper function to plot scatter plots, colored by cluster labels
plot_feature_pairs(clustered_wine_data, feature_names, color_labels=clustered_wine_data['Cluster'], title_prefix='Clustered Data: ')
```

We chose `k=3` for the number of clusters arbitrarily. However, in a real-world scenario, it is important to determine the optimal number of clusters using appropriate methods.

#### **Question 5:** 
#### Bootstrapping 

Implement bootstrapping on the mean of `color intensity`. Generate 10000 bootstrap samples, calculate the mean for each sample, and compute a 90% confidence interval.

```python
#My Answer: # Calculate the mean color intensity
mean_color_intensity = wine_df["color_intensity"].mean()

# Set the random seed so I get the same results if I run the code again
np.random.seed(123)

# Create 10,000 bootstrap samples
bootstrap_means = np.array([
    np.mean(
        np.random.choice(
            wine_df["color_intensity"],
            size=len(wine_df),
            replace=True
        )
    )
    for _ in range(10000)
])

# Find the lower and upper bounds for a 90% confidence interval
lower_bound = np.percentile(bootstrap_means, 5)
upper_bound = np.percentile(bootstrap_means, 95)

# Print the results
print(f"Mean of Color Intensity: {mean_color_intensity}")
print(f"90% Confidence Interval of Mean Color Intensity: ({lower_bound}, {upper_bound})")
```

**Question:**
- Why do we use bootstrapping in this context? What does it help us understand about the mean?

My Answer: Bootstrapping gives me a way to see how much the mean color intensity could change due to sampling variation. Instead of relying only on the mean from the original dataset, I repeatedly resample the data with replacement and calculate a new mean each time. After doing this 10,000 times, I can look at the distribution of those means and get a better idea of how consistent the original mean is.

**Question:**
- What is the purpose of calculating the confidence interval from the bootstrap samples? How does it help us interpret the variability of the estimate?

My Answer: The confidence interval gives a range around the estimated mean rather than treating the sample mean as one exact value. In this case, the 90% bootstrap confidence interval shows the range containing the central 90% of the bootstrapped mean estimates. This helps show how much uncertainty there is in the estimate of the mean color intensity.

**Question:**

- Reflect on the variability observed in the bootstrapped means and discuss whether the mean of the color intensity appears to be a stable and reliable estimate based on the confidence interval and the spread of the bootstrapped means.

My Answer: 
The mean color intensity was approximately 5.06, and the 90% bootstrap confidence interval ranged from approximately 4.78 to 5.35. The bootstrap estimates therefore remain fairly close to the original mean, with the interval extending about 0.28–0.29 units on either side. This suggests that the mean color intensity is reasonably stable across the resampled datasets, although there is still some sampling variability. Overall, the confidence interval gives a useful range for where the population mean could reasonably fall based on the sample.

Additionally, something I found useful from this analysis is that the confidence interval describes the uncertainty in the mean, rather than the variability of the individual wine samples. Even though individual color intensity measurements may vary considerably, repeatedly resampling the dataset produced mean values concentrated within a much smaller range. This helps show why averaging multiple observations can provide a more stable estimate than looking at individual measurements.
